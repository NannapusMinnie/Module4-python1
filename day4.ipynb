```python
# improve Nested If
if 2 != 3:
    if 3 != 4:
        if 4 != 5:
            if 5 != 6:
                if 6 != 7:
                    print('yay')
            if 11 != 13:
                print('huh')

if (2 != 3) and (3 != 4) and (4 != 5):
    if (5 != 6) and (6 != 7):
        print('yay')
    if (11 != 13):
        print('huh')
```

```python
# Nested If
if 5 != 4 and 6 > 3:
    if 5 < 4:
        print('No')
    print('Yess') # under the 1st if. the 1st if is true so it prints yes
```

```python
# Else
if 4 > 5:
    print('A')
if 5 > 4:
    print('C')
else:
    print('B')

if 4 < 5:
    print('A')
if 5 < 4:
    print('C')
else:
    print('B')
```

```python
# Elif: condition after if and before else
if 2 < 3:
    print('A')
elif 3 > 4:
    print('B')
elif 4 > 1:     # true but isn't printed
    print('D')
else:
    print('C')
# it will print the first true or else
```

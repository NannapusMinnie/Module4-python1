# Set Operations
union

intersection

difference

complimentary (not)

```python
set_a = {1, 2, 3}
set_b = {2, 3, 4}

# UNION
print(set_a.union(set_b)) # not in-place operation
print(set_a.union(set_b).union({4, 5, 6}))
print(set_a|set_b|{5, 6})

print(set_a.union([2, 3, 4])) # union.() accepts everything that is iterable
#set_a | [2, 3, 4] # | only accepts sets
```

```python
# INTERSECTION
print(set_a.intersection(set_b))
print(set_a & set_b)
```

```python
# DIFFERENCE
print(set_a.difference(set_b)) # in set_a but not in set_b
print(set_b.difference(set_a)) # in set_b but not in set_a

print(set_a - set_b)
print(set_b - set_a)

# can't + * /
```

```python
# COMPLIMENTARY
universe = {1, 2, 3, 4, 5, 6, 7, 8, 9}
universe - set_a
```

Modify Set

```python
# add
set_a = {2, 3}
set_a.add(False)
set_a.add(True)
print(set_a)

set_b = {0, 2}
set_b.add(False)
print(set_b)

set_c = {False, 3} # False = 0
set_c.add(0)
print(set_c)
set_c.add(True+1)
set_c
```

```python
# remove
set_a = {1, 2, 3}
set_a.remove(True)
set_a

set_b = {1, True}
print(set_b)

print(hash(0), hash(False), hash(""), hash(tuple()))
# set and list are not hashable

set_c = {0, '', False}
set_c
```

```python
# pop
set_a = {1, 2, 3, True, 0}
print(set_a.pop())
set_a
```

```python
# update
set_a = {1, 2, 3}
set_a.update({4, 5})
set_a
```

```python
# MODIFY
animal_legs['cheetah'] = 3
print(animal_legs)

animal_legs['lion'] = 4
print(animal_legs)

# Update (can replace value and add new key)
animal_legs.update({'cheetah':7, 'fish':100, 'insect':8})
print(animal_legs)

# similar to modify list
list1 = [1, 2, 3]
list1[2] = 4 # replace 3 with 4
print(list1)
```

```python
# pop() = remove and give value of the element
animal_legs = {'human':2, 'horse':4, 'fish':0}
print(animal_legs.pop('human'))
animal_legs
```

```python
# popitem()
animal_legs = {'human':2, 'horse':4, 'fish':0}
print(animal_legs.popitem())
animal_legs
```

```python
animal_legs = {'human':2, 'horse':4, 'fish':0, False:0}

# in / not in
print('human' in  animal_legs)
print(0 in animal_legs)
print(False not in animal_legs)

# can't check value in dict (you can only check KEY
```

```python
# Loop by keys
animal_legs = {'human':2, 'horse':4, 'fish':0}
for key in animal_legs:
    print(key)
    print(animal_legs[key])
```

```python
# .items()
animal_legs = {'human':2, 'horse':4, 'fish':0}

print(animal_legs.items())

for item in animal_legs.items():
    print(item, type(item))

for key, value in animal_legs.items():
    print(key, value)
```

```python
# .keys()
animal_legs = {'human':2, 'horse':4, 'fish':0}

print(animal_legs.keys())

for item in animal_legs.keys():
    print(item)
```

```python
# .values()
animal_legs = {'human':2, 'horse':4, 'fish':0}

print(animal_legs.values())

for item in animal_legs.values():
    print(item)
```

```python
# Enumerate
animal_legs = {'human':2, 'horse':4, 'fish':0}


for index, element in enumerate(animal_legs):
    print(index, element)
```

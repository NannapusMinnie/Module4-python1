# SET

set = a container that stores unique values (an object can only appears once)

```python
empty_set = {} # not empty set (this is empty dictionary)
type(empty_set)

empty_set = set()
my_set = {1, 1, 2, 3}
print(my_set)
```

```python
set([1, 1, 2, 3])
set('apple') # can't index the elements in set
```

```python
# ZH
number_of_elements = int(input())
elements = list(map(int, input().split()))
index_of_element = int(input())
print(elements[index_of_element])
```

```python
# ZI
number_of_elements = int(input())
elements = list(map(int, input().split()))
number = int(input())

for index in range(number_of_elements):
    elements[index] += number

print(*elements)
```

```python
# ZJ
number_of_elements = int(input())
elements = list(map(int, input().split()))
even_index = []
for index in range(number_of_elements):
    if index % 2 == 0:
        even_index.append(elements[index])
print(*even_index)
```

```python
# ZK
number_of_elements = int(input())
elements = list(map(int, input().split()))
even_element = []
for index in range(number_of_elements):
    if elements[index] % 2 == 0:
        even_element.append(elements[index])
print(*even_element)
```

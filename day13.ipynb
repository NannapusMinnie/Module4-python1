```python
# 0 input
# 0 output
def greeting():
    '''
    this fuction will print hello 3 times
    '''
    print('hello')
    print('hello')
    print('hello')
    return

print(greeting())

# we use comment to document the fuction
```

```python
# 1 input
# 1 output
def greet_person(name):
    '''
    greet a person according to the input name

    args:

    name (str) - name to be greeted
    '''
    return (f'Hello, {name}') # if there's no return statement, it'll print None.

print(greet_person('Minnie'))
```

```python
# Multiple inputs
import math
def calculate_gcd(num1, num2, num3):
    gcd_result = math.gcd(num1, num2, num3)
    return gcd_result

print(calculate_gcd(6, 8, 9))
```

```python
def f(y, x = 1, z = 7): # this is parameter
    result = x + y + z
    return result

print(f(2, 4)) # use default value of z
print(f(2, 4, 6)) # 2, 4, 6 are arguments
print(f(2))

# we can't have a parameter with default value follow by a parameter WITHOUT default value eg. x=1, y (need to swap y and x=1)
```

```python
def f(x, y, z):
    result = x + y + z
    return result

# f(2, 1, x=1) error can't input 2 xs

# 2 and 1 are positional argument
# x = 1 is keyword argument
# positionl argument can't follow keyword argument

f(1, y=2, z=3)

# f(2, z=8) error cause we didn't declare y and y have no default value
```

```python
# *args
def add_numbers(a, b, c, d):
    return a + b + c + d
print(add_numbers(4, 5, 6, 7))


def time_numbers(*args):
    print(args, type(args))
    result = 1
    for number in args:
        result *= number
    return result
print(time_numbers())
print(time_numbers(1, 2, 3))


def time_number_2(a, b, *args):
    result = a * b
    print(args)
    for number in args:
        result *= number
    return result
print(time_number_2(1, 2, 3, 4))
```

```python
def time_number_2(a, b, *args, c=5):
    result = a * b
    print(args)
    for number in args:
        result *= number
    return result
print(time_number_2(1, 2, 3, 4, 5, 6, c=0))
```

```python
# **kwargs
def print_names(**kwargs):
    print(type(kwargs))
    return kwargs

print(print_names())
print(print_names(x=5, y=6))
```

```python
def my_fuction(a, b, c=1, d=5, *args, e=8, **kwargs):
    # a and b are required parameters
    # c and d are parameters with default value
    print(a, b, c, d, args, e, kwargs)

my_fuction(1, 5, 6, 8, 10, 7, f=6, g=0)
# a=1, b=5, c=6, d=8, 10 and 7 in args, f=6 and g=10 in kwargs
```

```python
def my_fuction(a, b, c=1, *args, d=5, e=8, **kwargs):
    print(a, b, c, d, args, e, kwargs)

my_fuction(1, 5, 6, 8, 10, 7, f=6, g=0)
```

# Literal

```python
def add_numbers(a:int, b:int, c:int) -> int:
    return a + b + c
print(add_numbers(4, 5, 6))

def divide_numbers(a:int, b:int, c:int) -> float:
    return (a/b) / c
print(divide_numbers(4, 5, 6))

def square(x: float=7):
    return x ** 2
print(square(4))
print(square())
```

```python
word = input()
palindrome = True

for i in range(len(word) // 2):
    if word[i] != word[len(word) - 1 - i]:
        palindrome = False
        break

if palindrome:
    print("YES")
else:
    print("NO")
```
